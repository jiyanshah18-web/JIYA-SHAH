```python
countries = soup.select(".country")
len(countries)
```

```python
for country in countries:
    name = country.select_one(".country-name").get_text(strip=True)
    print(name)
```

```python
# Create an empty list.
# We'll store one dictionary per country in this list.
rows = []
# On the first iteration, 'country' is the first country.
# On the second iteration, it's the second country, and so on.
for country in countries:

    # Find the country name.
    country_name = country.select_one(".country-name")

    # Add one dictionary to our list.
    # Each dictionary becomes one row in our final dataframe.
    rows.append({

        # The name of the country.
        "country_name": country_name.get_text(strip=True),

        # Find the capital city and remove extra spaces.
        "capital": country.select_one(
            ".country-capital"
        ).get_text(strip=True),

        # Find the population and remove extra spaces.
        "population": country.select_one(
            ".country-population"
        ).get_text(strip=True),

        # Find the country's area and remove extra spaces.
        "area": country.select_one(
            ".country-area"
        ).get_text(strip=True),

        # There are no individual country links on this webpage,
        # so we store the main webpage URL.
        "country_url": url
    })

# Count how many dictionaries (countries) we collected.
len(rows)
```

```python
rows[0]
```

```python
df = pd.DataFrame(rows)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df["population"] = (
    df["population"]
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["area"] = (
    df["area"]
    .str.replace(",", "", regex=False)
    .astype(float)
)

df[["country_name", "capital", "population", "area"]].head()
```

```python
def data_summary(df):
    print("Rows:", len(df))

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nPopulation range:")
    print(df["population"].min(), "to", df["population"].max())

    print("\nArea range:")
    print(df["area"].min(), "to", df["area"].max())
```

```python
import time
import requests
import pandas as pd
from bs4 import BeautifulSoup

# Website URL
page_url = "https://www.scrapethissite.com/pages/simple/"

# Get the webpage
page_response = requests.get(page_url, timeout=30)
page_response.raise_for_status()

# Parse the webpage
page_soup = BeautifulSoup(
    page_response.text,
    "html.parser"
)

# Find all countries
page_countries = page_soup.select("div.country")

# Store the data
all_rows = []

for country in page_countries:

    all_rows.append({
        "country_name": country.select_one(
            ".country-name"
        ).get_text(strip=True),

        "capital": country.select_one(
            ".country-capital"
        ).get_text(strip=True),

        "population": country.select_one(
            ".country-population"
        ).get_text(strip=True),

        "area": country.select_one(
            ".country-area"
        ).get_text(strip=True)
    })

# Create DataFrame
country_df = pd.DataFrame(all_rows)

# Show the size of the dataset
print(country_df.shape)

# Show first 10 rows
country_df.head(10)
```
