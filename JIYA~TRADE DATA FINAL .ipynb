The code defines a function to calculate the percentage change between an old value and a new value.

```python
def percent_change(old, new):
    change = ((new - old) / old) * 100
    return change
```

```python
df["percent_change"] = percent_change(df["jul'25"], df["jul'26"])
```

The code defines a function to calculate percentage change, applies it to values of 100 and 125, and prints the resulting 25% change.

```python
def percent_change(old, new):
    change = ((new - old) / old) * 100
    return change

result = percent_change(100, 125)
print("Stored value:", result)
```

The code defines a function that identifies whether the value has increased, decreased, or remained unchanged.

```python
def direction(old, new):
    change = percent_change(old, new)
    if change > 0:
        return "increase"
    elif change < 0:
        return "decrease"
    else:
        return "no change"
```

```python
result = direction(100, 125)
print("Direction:", result)
```

**Story Pitch: India’s Changing Commodity Basket**


India’s commodity trade data points to a sharp increase in demand across sectors. The average value of the commodities in the dataset increased from 2,054 in July 2025 to 2,426 in July 2026, indicating an overall rise. 

However, the growth was not consistent. Electronic goods and cotton also rose by around 46%. In contrast, very few commodities moved in the opposite direction, with silver falling 66.1% and pulp and waste paper declining by 22.61%. The story will explore how the rise in imports highlights India’s continued dependence on other countries for a wide range of goods. 

Over just one year, several commodity categories recorded significant increases in import values, with some showing particularly sharp jumps. This raises questions about what is driving this growing demand for imported goods and what it means for India’s domestic production and economic dependence on global markets.

```python
import matplotlib.pyplot as plt

plot_df = df.sort_values("jul'26", ascending=False).head(10)

plt.figure(figsize=(12, 7))

x = range(len(plot_df))

plt.bar(
    [i - 0.2 for i in x],
    plot_df["jul'25"],
    width=0.4,
    label="July 2025",
    color="#38b6ff"
)

plt.bar(
    [i + 0.2 for i in x],
    plot_df["jul'26"],
    width=0.4,
    label="July 2026",
    color= "#ff66c4"
)

plt.xticks(x, plot_df["commodity"], rotation=45, ha="right")

plt.xlabel("Commodity")
plt.ylabel("Import Value")
plt.title("India's Top 10 Import Commodities: July 2025 vs July 2026")
```
